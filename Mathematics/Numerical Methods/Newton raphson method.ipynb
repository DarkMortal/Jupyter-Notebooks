# Newton raphson method

The Newton-Raphson method is based on the idea of linear approximation.

Suppose we want to find a root of a function $f(x)$, i.e., a value $x$ such that $f(x) = 0$.

We first need an initial guess. We'll call this guess value $x_1$. Let $f(x_1)=y_1$.

$\therefore$ Equation of the tangent to $f(x)$ at $x_1$ is
$$
(y-y_1)=f'(x)(x-x_1)
$$

Since we want to find $f(x)=0$, we'll replace $y=0$ and $y_1=f(x_1)$.

$\therefore -f(x_1)=f'(x_1)(x-x_1)\\\implies \boxed{x=x_1-\dfrac{f(x_1)}{f'(x_1)}}$
***


```python
import numpy as np

def newton_raphson(f, df, x0, tol=1e-6, max_iter=100):
    """
    Finds a root of a function using the Newton-Raphson method.

    Args:
        f (function): The function for which to find the root.
        df (function): The derivative of the function f.
        x0 (float): The initial guess for the root.
        tol (float): The tolerance for convergence.
        max_iter (int): The maximum number of iterations.

    Returns:
        float: The approximated root.
        None: If the method does not converge within max_iter.
    """
    x = x0
    for i in range(max_iter):
        f_x = f(x)
        df_x = df(x)

        if abs(df_x) < 1e-10:  # Avoid division by zero
            print("Derivative is too small, cannot proceed.")
            return None

        x_new = x - f_x / df_x

        if abs(x_new - x) < tol:
            print(f"Converged in {i+1} iterations.")
            return x_new
        x = x_new

    print("Maximum number of iterations reached.")
    return None
```

## Example
Let's find the root of the function $f(x) = x + \ln(x)$ whose derivative is $f'(x)=1+\dfrac{1}{x}$.

```python
from math import log

# Define the function and its derivative
def f(x):
    return x + log(x)

def df(x):
    return 1 + (1 / x)

# Initial guess
x0 = 2

# Find the root
root = newton_raphson(f, df, x0)

if root is not None:
    print(f"The approximated root is: {root:.6f}")
    print(f"Function value at the root: f({root:.6f}) = {f(root):.3f}")
```
