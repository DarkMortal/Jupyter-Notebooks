# Secant Method

A secant line of a curve is a line that intersects the curve at least at 2 distinct points.

We already know the Newton-raphson method
$$
\boxed{x_{n+1}=x_n-\dfrac{f(x_n)}{f'(x_n)}}
$$
Substituting
$$
f'(x_n)=\dfrac{y_n-y_{n-1}}{x_n-x_{n-1}}
$$
we get
$$
x_{n+1}=x_n-f(x)\dfrac{x_n-x_{n-1}}{y_n-y_{n-1}}
$$

Substituting $y_n=f(x_n)$ and $y_{n+1}=f(x_{n+1})$
we get
$$
\boxed{x_{n+1}=x_n-f(x)\dfrac{x_n-x_{n-1}}{f(x_n)-f(x_{n-1})}}
$$
***

```python
from math import log

def secant_method(f, x0, x1, tol=1e-6, max_iter=100):
    """
    Finds a root of a function using the Secant method.

    Args:
        f (function): The function for which to find the root.
        x0 (float): The first initial guess for the root.
        x1 (float): The second initial guess for the root.
        tol (float): The tolerance for convergence.
        max_iter (int): The maximum number of iterations.

    Returns:
        float: The approximated root.
        None: If the method does not converge within max_iter.
    """
    for i in range(max_iter):
        f_x0 = f(x0)
        f_x1 = f(x1)

        if abs(f_x1 - f_x0) < 1e-10:  # Avoid division by zero
            print("Denominator is too small, cannot proceed.")
            return None

        x_new = x1 - f_x1 * (x1 - x0) / (f_x1 - f_x0)

        if abs(x_new - x1) < tol:
            print(f"Converged in {i+1} iterations.")
            return x_new

        x0 = x1
        x1 = x_new

    print("Maximum number of iterations reached.")
    return None
```

## Application of Secant Method
Let's apply the Secant method to find the root of $f(x) = x + \ln(x)$.

```python
# Define the function
def f(x):
    return x + log(x)

# Initial guesses
x_initial_0 = 0.5  # Must be > 0 due to log(x)
x_initial_1 = 1.0

# Find the root using Secant method
root_secant = secant_method(f, x_initial_0, x_initial_1)

if root_secant is not None:
    print(f"The approximated root using Secant method is: {root_secant:.6f}")
    print(f"Function value at the root: f({root_secant:.6f}) = {f(root_secant):.3f}")
```
